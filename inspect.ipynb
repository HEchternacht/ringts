# Database Inspection Notebook

This notebook connects to your database and allows you to inspect its tables, data, and schema interactively.

## 1. Import Required Libraries
Import pandas and SQLAlchemy for database interaction and data inspection.

```python
import pandas as pd
from sqlalchemy import create_engine, inspect
```

## 2. Set Up Database Connection

Create a SQLAlchemy engine to connect to your database. Update the connection string below if needed.

```python
# Example for SQLite (update path if needed)
db_path = 'var/data/ringts.db'  # Update if your DB file is elsewhere
engine = create_engine(f'sqlite:///{db_path}')
```

## 3. Inspect Tables: List All Table Names

Use SQLAlchemy's inspect module to list all tables in the connected database.

```python
inspector = inspect(engine)
table_names = inspector.get_table_names()
print('Tables in database:')
for name in table_names:
    print('-', name)
table_names
```

## 4. Preview Table Data: Show First Rows

Read and display the first few rows of each table using pandas.

```python
# Display first 5 rows of each table
for name in table_names:
    print(f'\nTable: {name}')
    try:
        df = pd.read_sql_table(name, engine)
        display(df)
    except Exception as e:
        print(f'Could not read table {name}:', e)
```

## 5. Query Table Schema

Display the schema (column names and types) for each table.

```python
for name in table_names:
    print(f'\nSchema for table: {name}')
    try:
        columns = inspector.get_columns(name)
        for col in columns:
            print(f"- {col['name']}: {col['type']}")
    except Exception as e:
        print(f'Could not inspect table {name}:', e)
```

## 6. Run Custom SQL Queries

Use the cell below to run any custom SQL query and display the results as a DataFrame.

```python
# Example: Replace with your own query
query = "SELECT * FROM your_table LIMIT 10"
try:
    df = pd.read_sql_query(query, engine)
    display(df)
except Exception as e:
    print('Error running query:', e)
```
